```python
## sql- day1
##count how many customers are there in  each country
##find the number of customers from each country
##find the number of customers from each country who are older than 25yrs old

SELECT country, COUNT(*) AS customer_count
FROM customers
GROUP BY country;


##find the number of customers from each country
find the first name,last name age from customers  who are younger than 28yrs old

SELECT first_name, last_name, age
FROM customers
WHERE age < 28;

find the one who are younger than 28yrs and from UK

SELECT first_name, last_name, age, country
FROM customers
WHERE age < 28
  AND country = 'UK';
```

```python
What is a primary key
uniquely identify in a dataset

##fimd the number of customers
##find the number of customers for each country

select country,count(country) from customers group by country

## find the number of customers from each country who are older than 25yrs old
SELECT country, COUNT(*) AS count_22
FROM customers
WHERE age = 22
GROUP BY country;

## find the items and their amounts ,order id which are cheapers than 500

select order_id,amount from orders where amount<500

##get the avg age of customers group by country,avg(age)
select country,avg(age) from customers group by country

SELECT country, AVG(age) AS avg_age
FROM customers
GROUP BY country;

##count customers with the same last name
select last_name,count(last_name) from customers group by last_name

SELECT last_name, COUNT(*) AS count_last_name
FROM customers
GROUP BY last_name;

##find countries having more than 1 customer
select country,count(country) from customers group by country having count(country)>1

SELECT country, COUNT(*) AS customer_count
FROM customers
GROUP BY country
HAVING COUNT(*) > 1;


##count how many customers are aged 22 grouped by country
select country,count(country) from customers where age=22 group by country

```

```python
##find the total order amount for each item
SELECT item_id, SUM(amount) AS total_amount
FROM orders
GROUP BY item_id;

##count how many times each item was ordered
SELECT item_id, COUNT(*) AS times_ordered
FROM orders
GROUP BY item_id;

##get total amount spent by each customer
SELECT customer_id, SUM(amount) AS total_spent
FROM orders
GROUP BY customer_id;

##find customers who have placed more than 1 order(group by +having count)
SELECT customer_id, COUNT(*) AS order_count
FROM orders
GROUP BY customer_id
HAVING COUNT(*) > 1;

##cheaper than 500

SELECT order_id, item_id, amount
FROM orders
WHERE amount < 500;

##find items where total sales amount is greater than 500
SELECT item_id, SUM(amount) AS total_sales
FROM orders
GROUP BY item_id
HAVING SUM(amount) > 500;

##count number of orders where amount is greater than 300 grouped by item
SELECT item_id, COUNT(*) AS high_value_orders
FROM orders
WHERE amount > 300
GROUP BY item_id;
```

```python
##Count number of shipments by status
SELECT status, COUNT(*) AS shipment_count
FROM shipments
GROUP BY status;

##Count number of pending shipments.
SELECT COUNT(*) AS pending_shipments
FROM shipments
WHERE status = 'Pending';

##Count shipments per customer.
SELECT customer_id, COUNT(*) AS shipment_count
FROM shipments
GROUP BY customer_id;

##Find customers who have more than 1 shipment.
SELECT customer_id, COUNT(*) AS shipment_count
FROM shipments
GROUP BY customer_id
HAVING COUNT(*) > 1;

##Count delivered shipments grouped by customer.
SELECT customer_id, COUNT(*) AS delivered_count
FROM shipments
WHERE status = 'Delivered'
GROUP BY customer_id;

##Find status which appears more than once.
SELECT status, COUNT(*) AS status_count
FROM shipments
GROUP BY status
HAVING COUNT(*) > 1;

What is the difference between having clause and where clause ?
```
