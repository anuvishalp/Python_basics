```python
## Modules in Python
##Numpy
NumPy is the foundation of numerical computing in Python. It provides:

        Fast n‑dimensional arrays
        Vectorized operations (no Python loops)
        Linear algebra, statistics, random sampling
        Broadcasting for efficient computation

📌 Summary Table
Expression	Meaning	Output
arr[1,2]	Element at row 1, col 2	22
arr[3,1]	Element at row 3, col 1	41
arr[0]	Entire row 0	[10 11 12 13]
arr[:,3]	All rows, column 3	[13 23 33 43]
arr[:,1]	All rows, column 1	[11 21 31 41]
arr[:2,:2]	First 2 rows × first 2 columns	[[10 11],[20 21]]

"""
slicing with steps
boolean indexing
fancy indexing
multidimensional slicing
reshaping + indexing patterns """

🔹 1. Basic Indexing -> Works like Python lists, but extended to multiple dimensions.

arr[2]        # row 2
arr[2, 3]     # row 2, column 3
arr[-1]       # last row
arr[-1, -2]   # last row, second last column

🔹 2. Slicing Syntax -> general form:  start : stop : step

arr[1:4]      # rows 1,2,3
arr[:3]       # rows 0,1,2
arr[2:]       # rows 2 to end
arr[::2]      # every 2nd row
arr[::-1]     # reverse rows

🔹 3. Slicing With Steps->Works on rows, columns, or both.

➤ Every 2nd row -> arr[::2, :]
➤ Every 2nd column ->  arr[:, ::2]
➤ Reverse rows + reverse columns  -> arr[::-1, ::-1]
➤ Extract a checkerboard pattern->  arr[::2, ::2]

🔹 4. Multidimensional Slicing ->You can slice rows and columns simultaneously.

➤ Top‑left 3×3 block -> arr[:3, :3]
➤ Middle block -> arr[1:3, 1:4]
➤ Entire column -> arr[:, 2]
➤ Entire row -> arr[3, :]
➤ Submatrix with steps
    arr[1:4:2, 0:4:2]

🔹 5. Boolean Indexing ->Select elements based on conditions.

➤ Elements greater than 10 -> arr[arr > 10]
➤ Even numbers ->  arr[arr % 2 == 0]
➤ Multiple conditions (use &, |)
    arr[(arr > 10) & (arr < 30)]

➤ Boolean mask for rows

      mask = np.array([True, False, True, False])
      arr[mask]

➤ Boolean mask for columns -> arr[:, mask]

🔹 6. Fancy Indexing (Index Arrays) -> Use lists/arrays of indices.

➤ Select specific rows ->  arr[[0, 2, 3]]
➤ Select specific columns->  arr[:, [1, 3]]
➤ Select specific (row, col) pairs

arr[[0, 1, 2], [1, 2, 3]]
This picks:

      arr[0,1]

      arr[1,2]

      arr[2,3]

➤ Repeating indices -> arr[[1, 1, 1], [0, 2, 3]]
🔹 7. Fancy Indexing + Slicing
You can mix them.

➤ Select rows 0 and 2, but only columns 1:3 ->   arr[[0, 2], 1:3]
➤ Select columns 0 and 3 from rows 1:4 ->  arr[1:4, [0, 3]]

🔹 8. Reshaping + Indexing Patterns->Reshape changes the view; indexing works on the reshaped array.

➤ Flatten to 1D
    flat = arr.ravel()
    flat[5]
➤ Reshape to 2×8
    b = arr.reshape(2, 8)
    b[1, 3]
➤ Reshape + slice -> arr.reshape(8, 2)[:3]
➤ Convert to column vector -> arr.reshape(-1, 1)
➤ Convert to row vector -> arr.reshape(1, -1)

🔹 9. Indexing With np.newaxis / None -> Used to expand dimensions.

➤ Convert 1D → column vector -> x[:, None]
➤ Convert 1D → row vector -> x[None, :]
➤ Broadcasting trick -> arr[:, None, :]   # add a new axis in the middle

🔹 10. Advanced Patterns (Interview Favorites)
✔ Extract diagonal -> np.diag(arr)
✔ Extract anti‑diagonal -> np.diag(np.fliplr(arr))
✔ Mask upper triangle -> arr[np.triu_indices_from(arr)]
✔ Mask lower triangle -> arr[np.tril_indices_from(arr)]
✔ Replace values using boolean mask -> arr[arr < 20] = 0
```

```python
🌟 1. NumPy + Pandas Combined Cheatsheet (Ultimate Edition)
Below is a side‑by‑side, scenario‑driven cheatsheet showing how NumPy and Pandas solve similar problems differently.

🧱 A. Data Structures
Feature---------	NumPy	------------------Pandas
Primary object---------ndarray---------	Series, DataFrame
Data type	---------Homogeneous---------	Heterogeneous
Best for---------	Numerical computing----	Tabular data, analytics
Missing values---------	np.nan---------	NaN, None, nullable dtypes

Example
import numpy as np
import pandas as pd

np_arr = np.array([[1,2,3],[4,5,6]])
df = pd.DataFrame(np_arr, columns=['A','B','C'])

🧱 B. Creating Data
NumPy
  np.zeros((2,3))
  np.arange(0,10,2)
  np.random.randn(3,3)

Pandas
  pd.DataFrame({'A':[1,2], 'B':[3,4]})
  pd.read_csv('file.csv')

🧱 C. Indexing & Slicing
NumPy
  a[0:2, 1:]
  a[a > 5]
Pandas
  df.loc[0:2, 'A':'C']
  df[df['A'] > 5]
  df.iloc[0:3, 1:]

🧱 D. Aggregations
NumPy
    np.mean(a)
    np.sum(a, axis=0)
Pandas
    df.mean()
    df.groupby('col').sum()
    df.agg(['mean','std'])

🧱 E. Joining / Merging
NumPy
    np.concatenate([a1, a2], axis=0)
Pandas
    pd.merge(df1, df2, on='id')
    df1.join(df2)

🧱 F. Missing Values
NumPy
    np.isnan(arr)
    np.nanmean(arr)
Pandas
    df.isna()
    df.fillna(0)
    df.dropna()

🧱 G. Reshaping
NumPy
    a.reshape(2,3)
    a.ravel()
    a.T
Pandas
    df.pivot(index='A', columns='B', values='C')
    df.melt()
    df.T
========================================
🌟 2. NumPy Interview Questions + Answers

Q1. What is NumPy and why is it fast?
Answer:  NumPy is a numerical computing library built on top of optimized C code. It is fast because:

        Uses contiguous memory blocks
        Performs vectorized operations (no Python loops)
        Uses SIMD and BLAS/LAPACK under the hood
        Avoids Python object overhead

Q2. What is broadcasting?
Answer:  Broadcasting allows NumPy to perform operations on arrays of different shapes by expanding dimensions when compatible.

Example:

a = np.array([[1,2,3],[4,5,6]])
b = np.array([10,20,30])
a + b

Q3. Difference between view and copy?
Answer:

view → shares memory

copy → new memory allocation

      a = np.arange(5)
      b = a.view()
      c = a.copy()

Q4. What is vectorization?
Answer:  Replacing Python loops with optimized C‑level operations.

python
# slow
[x**2 for x in range(100000)]

# fast
np.arange(100000)**2

Q5. Explain np.where() -> Used for conditional selection.

    np.where(a > 5, 'big', 'small')

Q6. Explain axis parameter

      axis=0 → column‑wise
      axis=1 → row‑wise

Q7. Difference between NumPy and Pandas?

      NumPy → numerical arrays
      Pandas → labeled, tabular data

Q8. What is memory layout (C vs F order)?

        order='C' → row‑major

        order='F' → column‑major

Q9. How does NumPy handle missing values?-> Using np.nan and np.isnan().

Q10. What is np.vectorize()? ->A convenience wrapper, not true vectorization.
=========================================================================
🌟 3. NumPy Real‑World Scenarios (ETL, ML, Analytics)
Here are practical, industry‑grade examples.

🏗 A. ETL Scenarios
1. Cleaning numeric columns
        arr = np.array([1,2,np.nan,4])
        clean = np.nan_to_num(arr, nan=0)
2. Detecting outliers
      z = (arr - arr.mean()) / arr.std()
      outliers = arr[z > 3]
3. Converting data types -> arr.astype(np.float32)
4. Fast transformations
      scaled = (arr - arr.min()) / (arr.max() - arr.min())
========================================================================
🤖 B. Machine Learning Scenarios

1. Feature scaling -> X = (X - X.mean(axis=0)) / X.std(axis=0)
2. Distance calculation ->   dist = np.linalg.norm(a - b)
3. Gradient computation ->   grad = 2 * (pred - y) / len(y)
4. One‑hot encoding ->     np.eye(4)[labels]

====================================
📊 C. Analytics Scenarios
1. Rolling window->   np.convolve(arr, np.ones(3)/3, mode='valid')
2. Weighted average -> np.average(arr, weights=w)
3. Correlation -> np.corrcoef(a, b)
======================================
🌟 4. NumPy vs Pandas vs Python Lists (Full Comparison)
Feature	------Python List	--------NumPy--------	Pandas
Speed------	Slow      ------	Fast------------	Fast
Data type	------Mixed	------Homogeneous	------Mixed
Memory	------High------	Low------------	Medium
Operations------	Loops------	Vectorized	------Vectorized
Missing values------	None	------np.nan	------NaN, nullable
Best for	------General use------	Math, ML------	Analytics, ETL
===============================
🧪 Example: Square every element

          Python List ->    [x*x for x in lst]

          NumPy -> arr**2

          Pandas -> df['col']**2

🧪 Example: Filter values > 10

Python List -> [x for x in lst if x > 10]
NumPy -> arr[arr > 10]
Pandas -> df[df['col'] > 10]
---------------------------
🔥 A unified NumPy + Pandas PDF‑style study guide
🔥 Flashcards for quick revision
🔥 100 NumPy interview questions (basic → advanced)
🔥 NumPy → Pandas conversion table
=================================================
📘 1. Unified NumPy + Pandas Study Guide (PDF‑Style)

NUMPY + PANDAS STUDY GUIDE
🧱 1. Core Concepts
NumPy
    Numerical computing library
    Provides ndarray (homogeneous, fixed‑type)
    Fast due to C backend, vectorization, contiguous memory

Pandas
    Data analysis library
    Provides Series and DataFrame
    Built on top of NumPy
    Handles heterogeneous, labeled, tabular data

🧱 2. Data Structures

Feature	NumPy	Pandas
Primary object	ndarray	Series, DataFrame
Data type	Homogeneous	Mixed
Indexing	Integer-based	Labels + integer
Missing values	np.nan	NaN, nullable dtypes
Best for	Math, ML	ETL, analytics

🧱 3. Creating Data
NumPy
      np.array([1,2,3])
      np.zeros((3,3))
      np.arange(0,10,2)
      np.random.randn(4,4)
Pandas
      pd.Series([1,2,3])
      pd.DataFrame({'A':[1,2], 'B':[3,4]})
      pd.read_csv('file.csv')
🧱 4. Indexing & Slicing
NumPy
      a[1:4]
      a[:, 1:]
      a[a > 5]
Pandas
      df.loc[0:2, 'A':'C']
      df.iloc[0:3, 1:]
      df[df['A'] > 5]
🧱 5. Aggregations
NumPy
      np.mean(a)
      np.sum(a, axis=0)
      np.std(a)
Pandas
      df.mean()
      df.groupby('col').sum()
      df.agg(['mean','std'])
🧱 6. Reshaping
NumPy
      a.reshape(2,3)
      a.ravel()
      a.T
Pandas
      df.pivot(index='A', columns='B', values='C')
      df.melt()
      df.T
🧱 7. Merging / Joining
NumPy -> np.concatenate([a1, a2], axis=0)
Pandas ->
      pd.merge(df1, df2, on='id')
      df1.join(df2)
🧱 8. Missing Values
NumPy
      np.isnan(arr)
      np.nanmean(arr)
Pandas
      df.isna()
      df.fillna(0)
      df.dropna()
🧱 9. Performance Tips

Prefer vectorization
Avoid Python loops
Use astype() for dtype optimization
Use ravel() instead of flatten()
Use categorical dtype in Pandas for memory savings

📘 2. Flashcards for Quick Revision
Use these as Anki cards or quick mental drills.

Flashcard Set: NumPy
Q: What is NumPy’s main data structure?
A: ndarray — homogeneous, fixed‑type, n‑dimensional array.

Q: What is broadcasting?
A: Automatic expansion of shapes during arithmetic operations.

Q: Difference between view and copy?
A: View shares memory; copy creates new memory.

Q: How to generate random numbers?
A: np.random.rand(), np.random.randn(), np.random.randint().

Q: How to reshape an array?
A: a.reshape(r, c).

Flashcard Set: Pandas
Q: What is a Series?
A: 1D labeled array.

Q: What is loc vs iloc?
A: loc → label-based; iloc → integer-based.

Q: How to handle missing values?
A: df.fillna(), df.dropna(), df.isna().

Q: How to group data?
A: df.groupby('col').agg().

Q: How to merge DataFrames?
A: pd.merge(df1, df2, on='key').
============================================================================
📘 3. 100 NumPy Interview Questions (Basic → Advanced)

A. Basic (1–30)
What is NumPy?

What is an ndarray?

How do you create a NumPy array?

Difference between list and ndarray?

What is vectorization?

What is broadcasting?

How do you check array shape?

How do you reshape an array?

What is dtype?

How to convert dtype? (astype)

How to generate zeros/ones arrays?

How to generate random numbers?

What is np.arange()?

What is np.linspace()?

How to slice arrays?

How to index 2D arrays?

What is boolean indexing?

What is fancy indexing?

How to compute mean/sum/std?

What is axis?

How to concatenate arrays?

How to transpose?

How to flatten arrays?

Difference between flatten and ravel?

How to find unique values?

How to sort arrays?

How to find min/max?

How to find argmin/argmax?

How to check for NaN?

How to replace NaN?

B. Intermediate (31–70)
Explain broadcasting rules.

What is memory layout (C vs F order)?

What is np.where()?

What is np.select()?

How to compute correlation?

How to compute covariance?

What is np.dot()?

Difference between dot and matmul?

What is einsum()?

What is tile() vs repeat()?

How to stack arrays vertically/horizontally?

What is np.apply_along_axis()?

What is broadcasting with scalars?

How to normalize data?

How to standardize data?

How to compute Euclidean distance?

How to compute Manhattan distance?

What is np.linalg.inv()?

What is np.linalg.det()?

What is np.linalg.eig()?

What is np.linalg.svd()?

How to generate identity matrix?

How to clip values?

How to pad arrays?

How to reshape without knowing dimensions? (-1)

What is broadcasting error?

How to check memory usage?

How to compare arrays?

How to shuffle arrays?

How to set random seed?

What is np.vectorize()?

Why is vectorize not true vectorization?

How to compute rolling window?

How to compute weighted average?

How to compute histogram?

How to generate meshgrid?

What is np.fromiter()?

What is np.frombuffer()?

What is np.memmap()?

How to save/load arrays?

C. Advanced (71–100)
Explain broadcasting with 3D arrays.

What is stride?

How does NumPy use strides for slicing?

What is a view in terms of strides?

Explain contiguous vs non‑contiguous arrays.

What is ascontiguousarray()?

What is asfortranarray()?

What is np.lib.stride_tricks?

Explain sliding_window_view.

How does NumPy optimize memory?

What is BLAS/LAPACK?

How does NumPy use SIMD?

What is broadcasting with singleton dimensions?

Explain np.expand_dims().

Explain np.squeeze().

What is np.may_share_memory()?

What is np.bincount()?

What is np.digitize()?

What is np.percentile()?

What is np.quantile()?

What is np.gradient()?

What is np.diff()?

What is np.triu() and np.tril()?

What is np.kron()?

What is np.block()?

What is np.put() and np.take()?

What is np.choose()?

What is np.lexsort()?

What is np.msort()?

Explain broadcasting pitfalls in ML.
=================================================================
📘 4. NumPy → Pandas Conversion Table

Task	--------NumPy----------------	Pandas

Create array/table----	np.array()	pd.DataFrame()
Select rows----	a[1:3]	------------df.iloc[1:3]
Select columns----	a[:,1]	----df['col']
Boolean filter----	a[a>5]	----df[df['col']>5]
Replace values----	np.where()----	df.replace()
Missing values----	np.isnan()----	df.isna()
Fill missing----	np.nan_to_num()	----df.fillna()
Aggregation	----np.mean(a)----	df.mean()
Grouping----	manual loops	----df.groupby()
Merge	----np.concatenate()	----pd.merge()
Reshape	----reshape()----	pivot(), melt()
Stats	----np.std()	----df.describe()
```
