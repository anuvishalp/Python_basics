```python
import numpy as np

# Create a 1D array with values from 1 to 10
arr = np.arange(1, 11)

# Square each element
squared = arr ** 2

# Filter elements greater than 50
filtered = squared[squared > 50]

# Find mean and standard deviation
mean_val = filtered.mean()
std_val = filtered.std()

mean_val, std_val
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# --- Base DataFrame ---
data = {
    "Name": ["Alice", "Bob", "Charlie", "David", "Eve"],
    "Age": [25, 30, 35, 29, 33],
    "Salary": [55000, 60000, 52000, 58000, 44000],
    "Department": ["HR", "Marketing", "Sales", "Sales", "Marketing"]
}

df = pd.DataFrame(data)

# 1. Show avg age and maximum salary
avg_age = df["Age"].mean()
max_salary = df["Salary"].max()
print("Average Age:", avg_age)
print("Maximum Salary:", max_salary)
```

```python
print(df['Department'].value_counts())
```

```python
print(df["Salary"].max())
```

```python
# 2. Count employees in each department
dept_counts = df["Department"].value_counts()
print("\nEmployee Count per Department:")
print(dept_counts)
```

```python
# 3. Plot histogram of Age
plt.figure(figsize=(6,4))
plt.hist(df["Age"], bins=5, color="skyblue", edgecolor="black")
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
```

```python
df["Age"].plot(kind="hist",edgecolor="black")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.title("histogram of age")
plt.show
```

```python
# 4. Scatter plot of Age vs Salary
plt.figure(figsize=(6,4))
plt.scatter(df["Age"], df["Salary"], color="green")
plt.title("Age vs Salary")
plt.xlabel("Age")
plt.ylabel("Salary")
plt.show()
```

```python
# 5. Bar chart of employee count per department
plt.figure(figsize=(6,4))
dept_counts.plot(kind="bar", color="orange", edgecolor="black")
plt.title("Employees per Department")
plt.xlabel("Department")
plt.ylabel("Count")
plt.show()
```

```python
df['Department'].value_counts().plot(kind='bar')
plt.xlabel('Department')
plt.ylabel('Count')
plt.title('Employees per Department')
plt.show()
```
