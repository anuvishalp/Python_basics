```python
##python Tutorial - Data structures

🧩 High‑Level Differences (One‑Glance Table)
Feature	List	Tuple	Set	Dictionary
Ordered	✔	✔	✘	✔ (Python 3.7+)
Mutable	✔	✘	✔	✔
Allows duplicates	✔	✔	✘	Keys ✘, Values ✔
Indexable	✔	✔	✘	Keys act as indexes
Use-case	Collections that change	Fixed collections	Unique items, fast lookup	Key–value mapping
==========================================

🟦 LIST -> Ordered • Mutable • Allows duplicates • Indexable
When to use

        When you need a dynamic, growable/shrinkable collection.
        When order matters.
        When you need to store duplicates.
        When you need slicing, indexing, sorting.

Common Scenarios
1. Dynamic data (shopping cart)
          cart = ["apple", "banana"]
          cart.append("orange")
          cart.remove("banana")
          print(cart)
2. Sorting and slicing
        nums = [5, 2, 9, 1]
        nums.sort()
        print(nums[:2])   # slicing
3. List of dictionaries (API responses)
        users = [
            {"id": 1, "name": "Anu"},
            {"id": 2, "name": "Jyoti"}
        ]

LIST Cheat‑Sheet (All Important Methods)
        append(x)
        extend(iterable)
        insert(i, x)
        remove(x)
        pop(i)
        clear()
        index(x)
        count(x)
        sort()
        reverse()
        copy()
================================
🟧 TUPLE ->Ordered • Immutable • Allows duplicates • Indexable

When to use

        When data should not change.
        When you want hashable items (e.g., keys in dict).
        When representing fixed structures (coordinates, RGB, DB rows).

Common Scenarios
1. Coordinates / fixed-size data -> point = (10, 20)

2. Returning multiple values from a function

        def stats(nums):
            return (min(nums), max(nums))

        mn, mx = stats([3, 9, 1])

3. Using tuples as dictionary keys
        distances = {
            ("Seattle", "Bellevue"): 12,
            ("Bellevue", "Redmond"): 8
        }

Tuples have very few methods (because immutable):

        count(x)
        index(x)
======================================================
🟩 SET -> Unordered • Mutable • Unique elements • No indexing

When to use
      When you need unique items.
      When you need fast membership tests (O(1)).
      When performing set operations (union, intersection, difference).

Common Scenarios
1. Removing duplicates
        nums = [1, 2, 2, 3, 3, 3]
        unique = set(nums)
2. Membership testing
        allowed = {"admin", "editor"}
        print("admin" in allowed)
3. Set operations (analytics)
        students_A = {"Anu", "Ravi", "Kiran"}
        students_B = {"Kiran", "Sita"}

        print(students_A & students_B)  # intersection
        print(students_A | students_B)  # union
        print(students_A - students_B)  # difference

SET Cheat‑Sheet (All Important Methods)
          add(x)
          update(iterable)
          remove(x)
          discard(x)
          pop()
          clear()
          union(s)
          intersection(s)
          difference(s)
          symmetric_difference(s)
          issubset(s)
          issuperset(s)
          isdisjoint(s)
          copy()
================================================
🟨 DICTIONARY-> Ordered • Mutable • Keys unique • Values can duplicate
When to use
        When you need key–value mapping.
        When you need fast lookup by key.
        When representing structured data (JSON-like).

Common Scenarios
1. Storing user profiles
          user = {
              "name": "Anu",
              "role": "Architect",
              "location": "Bellevue"
          }
2. Counting occurrences (frequency map)
        text = "banana"
        freq = {}

        for ch in text:
            freq[ch] = freq.get(ch, 0) + 1
3. Nested dictionaries (API-like)
        employee = {
            "id": 101,
            "name": "Anu",
            "skills": ["Python", "ETL", "SQL"],
            "address": {"city": "Bellevue", "state": "WA"}
        }

DICTIONARY Cheat‑Sheet (All Important Methods)
          get(key, default)
          keys()
          values()
          items()
          update(other_dict)
          pop(key)
          popitem()
          setdefault(key, default)
          clear()
          copy()
          fromkeys(iterable, value)

🧠 When to Choose What (Decision Guide)
Need	Best Choice

1.
        Fixed data, safe from modification	Tuple
        Need hashable items (keys in dict)	Tuple

2.
Unique items, fast membership	Set
Need to remove duplicates	Set

3.
        Dynamic, ordered, duplicates allowed	List

4.
Key–value mapping	Dictionary

5.
        Need to preserve insertion order	List / Dict
```

```python
### Complete LAMBDA expression details-> Map(),Filter(),Reduce(),Sorted(),Apply()

🧠 Core idea in one line
A lambda is a function without a name, defined using a single expression, often used when you need a function temporarily or inline.

        lambda arguments: expression

It always returns the value of the expression.

1. What a lambda is and is not

✔️ A lambda is

          A function object (same as one created with def)
          Anonymous (can be assigned to a variable, but doesn’t need to)
          Limited to one expression
          Useful for short, throwaway logic

❌ A lambda is not
        A replacement for def
        Able to contain statements (return, if, for, while, try, etc.)
        Multi-line
==================================
2. Basic usage patterns
2.1 Assigning a lambda to a variable

      square = lambda x: x * x
      square(5)   # 25

2.2 Lambda with multiple arguments
        add = lambda a, b: a + b
        add(3, 4)   # 7

2.3 Lambda with no arguments
      hello = lambda: "Hi Anu!"
      hello()

3. Lambdas with built‑in higher‑order functions -> These are the most common real-world scenarios.

3.1 map() + lambda -> Transform each element.

Example: Convert list of strings to lengths

      names = ["Anu", "Jyoti", "Python"]
      lengths = list(map(lambda x: len(x), names))

Example: Convert Fahrenheit → Celsius
        temps_f = [32, 68, 100]
        temps_c = list(map(lambda f: (f - 32) * 5/9, temps_f))

3.2 filter() + lambda -> Keep only elements that satisfy a condition.

Example: Keep even numbers
      nums = [1,2,3,4,5,6]
      evens = list(filter(lambda x: x % 2 == 0, nums))

Example: Keep strings starting with “A”
      names = ["Anu", "Bala", "Arun"]
      starts_with_a = list(filter(lambda x: x.startswith("A"), names))

3.3 sorted() + lambda -> Custom sorting logic.

Example: Sort tuples by second element

      pairs = [(1, 3), (2, 1), (5, 2)]
      sorted_pairs = sorted(pairs, key=lambda x: x[1])

Example: Sort dictionary by value
      d = {"a": 3, "b": 1, "c": 2}
      sorted_items = sorted(d.items(), key=lambda x: x[1])

3.4 reduce() + lambda -> Requires functools.

Example: Multiply all numbers

      from functools import reduce
      product = reduce(lambda a, b: a * b, [1,2,3,4])

4. Lambdas inside data structures -> Useful for strategy patterns, dynamic behavior, switch-case alternatives.

Example: Dictionary of operations

        ops = {
            "add": lambda a, b: a + b,
            "sub": lambda a, b: a - b,
            "mul": lambda a, b: a * b
        }

        ops["mul"](3, 4)   # 12
===============================================================
5. Lambdas in Pandas (very important for data engineering)

5.1 apply() on Series
        df["name_length"] = df["name"].apply(lambda x: len(x))

5.2 apply() on DataFrame rows
        df["full_name"] = df.apply(lambda row: row["first"] + " " + row["last"], axis=1)

5.3 Conditional transformation
        df["category"] = df["score"].apply(lambda x: "High" if x > 80 else "Low")

6. Lambdas with default arguments -> Useful for capturing values at definition time.

        funcs = []
        for i in range(3):
            funcs.append(lambda x, i=i: x + i)

        [f(10) for f in funcs]   # [10, 11, 12]

7. Lambdas with inline conditional expressions-> Allowed because it’s still a single expression.

        grade = lambda s: "Pass" if s >= 50 else "Fail"
        grade(70)
===========================================
8. Lambdas inside classes
8.1 As small helper functions

        class Calculator:
            double = lambda self, x: x * 2

8.2 As key functions for sorting class objects
        students = [
            {"name": "Anu", "score": 95},
            {"name": "Bala", "score": 80}
        ]

        sorted_students = sorted(students, key=lambda s: s["score"])
===========================================
9. Advanced scenarios

9.1 Lambda returning another lambda (function factories)

            make_multiplier = lambda n: lambda x: x * n
            times3 = make_multiplier(3)
            times3(10)   # 30

9.2 Lambda inside list comprehensions
            funcs = [lambda x, n=n: x + n for n in range(5)]
            [f(10) for f in funcs]   # [10,11,12,13,14]

9.3 Lambda for custom key extraction in complex data
            employees = [
                {"name": "Anu", "dept": {"id": 2}},
                {"name": "Jyoti", "dept": {"id": 1}}
            ]

            sorted_emps = sorted(employees, key=lambda e: e["dept"]["id"])

10. When NOT to use lambda
❌ When logic is long
❌ When readability suffers
❌ When you need multiple statements
❌ When debugging is needed (lambdas have no name)

11. Lambda vs def — interview comparison table
Feature	lambda	def
Name	Anonymous	Named
Lines	Single expression	Multi-line
Statements allowed	No	Yes
Readability	Lower	Higher
Use case	Short, inline logic	Full functions
==============================================

12. Real-world patterns (Data Engineering + Interview Ready)

✔️ ETL transformations
        clean = lambda s: s.strip().lower()
        df["clean_name"] = df["name"].apply(clean)

✔️ Feature engineering
        df["is_long"] = df["text"].apply(lambda x: 1 if len(x) > 100 else 0)

✔️ Sorting logs by timestamp
        sorted_logs = sorted(logs, key=lambda x: x["timestamp"])

✔️ Grouping + custom aggregation

        df.groupby("category")["value"].agg(lambda x: x.max() - x.min())
```
