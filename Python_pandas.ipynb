```python
##🔹 Filtering (Boolean indexing)

df[df["Age"] > 25]
df[df["City"] == "New York"]

##Great for row-level conditions ->Can combine multiple conditions with & and | (wrapped in parentheses)

##🔹 Handling missing data

df.isnull()        # Boolean mask of missing values
df.dropna()        # Remove rows with any missing values
df.fillna(0)       # Replace missing values with 0

##Useful variations:

df.dropna(subset=["Age"])

df.fillna({"Age": df["Age"].mean()})

##🔹 Sorting
df.sort_values(by="Age")
df.sort_values(by="Age", ascending=False)

##You can sort by multiple columns:

df.sort_values(by=["City", "Age"])

##🔹 Grouping (aggregation)

df.groupby("Age").mean(numeric_only=True)
df.groupby("City").sum()

##More expressive patterns:

df.groupby("City")["Age"].mean()
df.groupby("City").agg({"Age": "mean", "Salary": "sum"})
```

```python
"""
⭐ Multiple Ways to Calculate Average in pandas
=======================================================

1️⃣ Column Mean (most common)

df["Age"].mean()

Computes the average of a single column.

df["Age"].mean()

2️⃣ DataFrame Mean (all numeric columns)
df.mean() ->Returns mean for every numeric column.

python
df.mean(numeric_only=True)
3️⃣ Group-wise Mean
df.groupby("City")["Age"].mean()
Average Age per City.

python
df.groupby("City")["Age"].mean()
Multiple aggregations:
python
df.groupby("City").agg({"Age": "mean", "Salary": "mean"})
4️⃣ Mean with agg()
df.agg("mean")
Applies mean across all numeric columns.

python
df.agg("mean")
Or for specific columns:

python
df.agg({"Age": "mean"})
5️⃣ Using describe() (mean included)
df["Age"].describe()
Gives count, mean, std, min, quartiles, max.

python
df["Age"].describe()
6️⃣ Weighted Average (custom)
Pandas doesn’t have a built‑in weighted mean, but you can compute it easily:

python
import numpy as np
np.average(df["Age"], weights=df["Salary"])
7️⃣ Mean after filtering
Average of a subset:

python
df[df["City"] == "New York"]["Age"].mean()
8️⃣ Mean ignoring missing values (default behavior)
Pandas automatically ignores NaN:

python
df["Age"].mean()
But you can enforce it:

python
df["Age"].mean(skipna=True)
9️⃣ Mean using .loc or .iloc
python
df.loc[:, "Age"].mean()
df.iloc[:, 3].mean()   # if Age is 4th column
🔟 Mean using apply()
(Not recommended for performance, but useful for teaching)

python
df.apply(lambda col: col.mean())
1️⃣1️⃣ Rolling Mean (moving average)
Useful for time series:

python
df["Age"].rolling(window=3).mean()
1️⃣2️⃣ Expanding Mean (cumulative average)
python
df["Age"].expanding().mean() """
```

```python
"""
1️⃣ df = pd.read_csv("data.csv")
Loads a CSV file into a pandas DataFrame.

Rows → records

Columns → fields

This is usually the first step in any data‑analysis workflow.

2️⃣ df.head()
Shows the first 5 rows.

Useful for: sanity‑checking the data
            confirming column names
            spotting obvious issues

You can also specify a number:

df.head(10)

3️⃣ df.tail() ->Shows the last 5 rows.

Great for: checking if the file loaded fully
          inspecting trailing values
          seeing if there are footer rows or junk at the end

4️⃣ df.info()
Displays:

column names

data types

number of non‑null values

memory usage

This is your data health report — perfect for spotting missing values or wrong data types.

5️⃣ df.duplicated() -> Returns a Boolean Series indicating which rows are duplicates.

Example output:

Code
0    False
1    False
2    True
3    False
6️⃣ df.drop_duplicates() -> Removes duplicate rows.

You can also specify columns:

df.drop_duplicates(subset=["Name", "Age"])

Or keep the last occurrence:

df.drop_duplicates(keep="last")

7️⃣ df.rename(columns={"old_name": "new_name"}) ->Renames one or more columns.

Example:

df.rename(columns={"Salary": "AnnualSalary"})

To rename multiple:

df.rename(columns={"Name": "EmployeeName", "Age": "EmployeeAge"}) """
```

```python
import pandas as pd

data = {
    "Name": ["Alice", "Bob", "Charlie", "David", "Eve"],
    "Subject": ["Maths", "Hindi", "English", "Science", "Accounts"],
    "Marks": [90, 85, 92, 88, 95]
}

df = pd.DataFrame(data)
df
```

```python
#Filter students who scored 80 or above
df[df["Marks"] >= 80]
```

```python
##2️⃣ Using query()

df.query("Marks >= 80")
```

```python
##3️⃣ Using loc
df.loc[df["Marks"] >= 80]
```

```python
##4️⃣ Filtering + selecting only names
df.loc[df["Marks"] >= 80, "Name"]
```

```python
##5️⃣ Filtering + selecting multiple columns
df.loc[df["Marks"] >= 80, ["Name", "Marks"]]
```
