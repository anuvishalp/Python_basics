```python
##🌱 1. Basic Setup
import matplotlib.pyplot as plt
import numpy as np
```

```python
🆚 Differences Between Graph Types (Summary Table)
Plot-> Type	Data-> Type	Best For->	Not Good For
Line---	Continuous---	Trends---	Categories
Scatter	---Paired points---	Correlation---	Time series
Bar	---Categorical	---Comparison---	Distributions
Histogram	---Continuous	---Distribution	---Categories
Box---	Continuous---	Spread/outliers	---Exact values
Violin---	Continuous---	Density	---Small data
Heatmap---	Matrix	---Correlation	---Sparse data
Pie	---Categorical---	Proportions---	Many categories
3D---	3D data---	Surfaces	---Simple plots
Step---	Discrete	---Signals---	Smooth trends
Stem	---Discrete	---Impulses	---Continuous data
Quiver	---Vectors	---Direction fields	---Simple charts
Stream	---Vectors	---Flow	---Discrete data
Hexbin	---Dense--- scatter---	Density---	Small data
```

```python
##Create sample data
x = np.linspace(0, 10, 100)
y = np.sin(x)
plt.plot(x, y)
plt.show()
```

```python
##📊 2. Core Plot Types -> Line Plot

plt.plot(x, y, label="sin(x)", color="blue", linewidth=2, linestyle="--")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.title("Line Plot Example")
plt.legend()
plt.show()
```

```python
##Scatter Plot

plt.scatter(x, y, c='red', s=40, alpha=0.7)
plt.show()
```

```python
##Bar Chart
categories = ["A", "B", "C"]
values = [10, 20, 15]
plt.bar(categories, values, color="teal")
```

```python
##Horizontal Bar
plt.barh(categories, values)
```

```python
##Histogram
data = np.random.randn(1000)
plt.hist(data, bins=30, color="purple", edgecolor="black")
```

```python
##Box Plot
plt.boxplot([data, data*1.2])
```

```python
##Pie Chart
plt.pie(values, labels=categories, autopct="%1.1f%%", explode=[0,0.1,0])
```

```python
##🧩 3. Subplots (All Scenarios) -> Simple 1×2

fig, ax = plt.subplots(1, 2, figsize=(10, 4))
ax[0].plot(x, y)
ax[1].scatter(x, y)
```

```python
## 2×2 Grid
fig, ax = plt.subplots(2, 2, figsize=(8, 6))
ax[0,0].plot(x, y)
ax[0,1].bar(categories, values)
ax[1,0].hist(data)
ax[1,1].pie(values)
```

```python
##Shared Axes
fig, ax = plt.subplots(2, 1, sharex=True)
ax[0].plot(x, np.sin(x))
ax[1].plot(x, np.cos(x))
```

```python
##Add subplot manually
fig = plt.figure()
ax1 = fig.add_subplot(2,2,1)
ax2 = fig.add_subplot(2,2,4)
```

```python
##🎨 4. Styling & Customization
""" Colors
"red", "blue", "green"

Hex: "#FF5733"

RGB: (0.2, 0.4, 0.6)

Line Styles
"-" solid

"--" dashed

"-." dash-dot

":" dotted

Markers
"o" circle

"s" square

"^" triangle

"*" star
"""
plt.plot(x, y, marker="o", markersize=6)
```

```python
##Grid
plt.grid(True, linestyle="--", alpha=0.5)
```

```python
##Figure Size
plt.figure(figsize=(10, 5))
```

```python
##Fonts
plt.title("Plot", fontsize=16, fontweight="bold")
```

```python
##🧭 5. Axes Control -> Limits

plt.xlim(0, 10)
plt.ylim(-1, 1)
```

```python
##Ticks
plt.xticks(np.arange(0, 11, 2))
plt.yticks([-1, 0, 1])
```

```python

##Log Scale
plt.yscale("log")
```

```python
##Twin Axes
fig, ax1 = plt.subplots()
ax2 = ax1.twinx()

ax1.plot(x, y, color="blue")
ax2.plot(x, np.cos(x), color="red")
```

```python
##🖼️ 6. Images-> Show Image

img = plt.imread("image.png")
plt.imshow(img)
plt.axis("off")
```

```python
##🧱 7. 3D Plots
from mpl_toolkits.mplot3d import Axes3D

fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

X = np.linspace(-5, 5, 50)
Y = np.linspace(-5, 5, 50)
X, Y = np.meshgrid(X, Y)
Z = np.sin(np.sqrt(X**2 + Y**2))

ax.plot_surface(X, Y, Z, cmap="viridis")
```

```python
##🎞️ 8. Animation (Most Common Pattern)

from matplotlib.animation import FuncAnimation

fig, ax = plt.subplots()
line, = ax.plot([], [])

def init():
    ax.set_xlim(0, 10)
    ax.set_ylim(-1, 1)
    return line,

def update(frame):
    x = np.linspace(0, 10, 100)
    y = np.sin(x + frame)
    line.set_data(x, y)
    return line,

ani = FuncAnimation(fig, update, frames=100, init_func=init)
plt.show()
```

```python
##🧾 11. Full Template (Reusable)
import matplotlib.pyplot as plt
import numpy as np

plt.style.use("seaborn-v0_8")

fig, ax = plt.subplots(figsize=(10, 5))

x = np.linspace(0, 10, 100)
y = np.sin(x)

ax.plot(x, y, label="sin(x)", color="blue", linewidth=2)
ax.set_xlabel("X-axis")
ax.set_ylabel("Y-axis")
ax.set_title("Matplotlib Template")
ax.legend()
ax.grid(True)

plt.tight_layout()
plt.show()
```
